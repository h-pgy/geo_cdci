```python
import requests

url_wfs = "https://wfs.geosampa.prefeitura.sp.gov.br/geoserver/geoportal/wfs"

coord_x = 333312.00
coord_y = 7394464.00
raio_metros = 50
coluna_geometria = "ge_poligono" 

parametros = {
    "service": "WFS",
    "version": "1.1.0",
    "request": "GetFeature",
    "typeName": "geoportal:lote_cidadao",
    "outputFormat": "application/json",
    "srsName": "EPSG:31983",
    "cql_filter": f"DWITHIN({coluna_geometria}, POINT({coord_x} {coord_y}), {raio_metros}, meters)"
}

resposta = requests.get(url_wfs, params=parametros)
dados = resposta.json()
```
